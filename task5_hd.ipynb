```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("cleaned_dataset.csv")
df
```

```python
numeric_col = df.select_dtypes(include=np.number).columns
numeric_col.tolist()
```

```python
df[numeric_col].describe()
```

```python
# Histogram of all numerical columns

for col in numeric_col:
    plt.figure(figsize=(8,5))

    plt.hist(df[col].dropna(),bins=20,edgecolor="black",alpha=0.75)
    plt.title(f"distribution of {col}" , fontsize=16,fontweight="bold")
    plt.xlabel(col,fontsize=12)
    plt.ylabel("frequency",fontsize=12)

    plt.grid(axis="y",alpha=0.3)
    plt.tight_layout()
    plt.show()
```

```python
# Mean aur Median ke saath Histogram
for col in numeric_col:
    data = df[col].dropna()
    
    mean_value = data.mean()
    median_value = data.median()
    
    plt.figure(figsize=(9, 5))
    
    plt.hist(
        data,
        bins=20,
        edgecolor="black",
        alpha=0.7
    )
    
    plt.axvline(
        mean_value,
        linestyle="--",
        linewidth=2,
        label=f"Mean = {mean_value:.2f}"
    )
    
    plt.axvline(
        median_value,
        linestyle="-",
        linewidth=2,
        label=f"Median = {median_value:.2f}"
    )
    
    plt.title(f"{col} Distribution with Mean & Median",
              fontsize=16,
              fontweight="bold")
    
    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.legend()
    plt.grid(axis="y", alpha=0.3)
    
    plt.tight_layout()
    plt.show()
```

```python
# Boxplot for Outlier Detection
for col in numeric_col:
    data = df[col].dropna()
    
    plt.figure(figsize=(8, 3))
    
    plt.boxplot(
        data,
        vert=False,
        patch_artist=True
    )
    
    plt.title(
        f"Boxplot of {col}",
        fontsize=16,
        fontweight="bold"
    )
    
    plt.xlabel(col)
    plt.grid(axis="x", alpha=0.3)
    
    plt.tight_layout()
    plt.show()
```

```python
# Multiple numerical columns ka comparative boxplot
plt.figure(figsize=(14, 7))

plt.boxplot(
    [df[col].dropna() for col in numeric_col],
    labels=numeric_col,
    patch_artist=True
)

plt.title(
    "Comparison of Numerical Column Distributions",
    fontsize=18,
    fontweight="bold"
)

plt.xlabel("Numerical Features")
plt.ylabel("Values")

plt.xticks(rotation=45)
plt.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# Standardized Boxplot for Fair Comparison
standardized_data = []

for col in numeric_col:
    data = df[col].dropna()
    
    mean = data.mean()
    std = data.std()
    
    standardized = (data - mean) / std
    standardized_data.append(standardized)

plt.figure(figsize=(14, 7))

plt.boxplot(
    standardized_data,
    labels=numeric_col,
    patch_artist=True
)

plt.title(
    "Standardized Distribution Comparison",
    fontsize=18,
    fontweight="bold"
)

plt.xlabel("Numerical Features")
plt.ylabel("Standardized Value (Z-score)")

plt.xticks(rotation=45)
plt.axhline(0, linestyle="--", linewidth=1)

plt.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# Histogram comparison in one figure
n = len(numeric_col)

fig, axes = plt.subplots(
    nrows=n,
    ncols=1,
    figsize=(10, 5 * n)
)

if n == 1:
    axes = [axes]

for ax, col in zip(axes, numeric_col):
    
    data = df[col].dropna()
    
    ax.hist(
        data,
        bins=20,
        edgecolor="black",
        alpha=0.75
    )
    
    mean_value = data.mean()
    median_value = data.median()
    
    ax.axvline(
        mean_value,
        linestyle="--",
        linewidth=2,
        label=f"Mean: {mean_value:.2f}"
    )
    
    ax.axvline(
        median_value,
        linestyle="-",
        linewidth=2,
        label=f"Median: {median_value:.2f}"
    )
    
    ax.set_title(
        f"{col} Distribution",
        fontsize=14,
        fontweight="bold"
    )
    
    ax.set_xlabel(col)
    ax.set_ylabel("Frequency")
    ax.legend()
    ax.grid(axis="y", alpha=0.3)

plt.suptitle(
    "Distribution of All Numerical Features",
    fontsize=20,
    fontweight="bold"
)

plt.tight_layout()
plt.show()
```

```python
# Cumulative distribution / ECDF using only Matplotlib

for col in numeric_col:
    data = np.sort(df[col].dropna())
    
    y = np.arange(1, len(data) + 1) / len(data)
    
    plt.figure(figsize=(8, 5))
    
    plt.plot(
        data,
        y,
        linewidth=2
    )
    
    plt.title(
        f"ECDF of {col}",
        fontsize=16,
        fontweight="bold"
    )
    
    plt.xlabel(col)
    plt.ylabel("Cumulative Probability")
    
    plt.grid(alpha=0.3)
    
    plt.tight_layout()
    plt.show()
```

```python
# Distribution + outlier annotation
for col in numeric_col:

    data = df[col].dropna()

    Q1 = data.quantile(0.25)
    Q3 = data.quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = data[
        (data < lower_bound) |
        (data > upper_bound)
    ]

    plt.figure(figsize=(9, 5))

    plt.hist(
        data,
        bins=20,
        edgecolor="black",
        alpha=0.7
    )

    plt.axvline(
        lower_bound,
        linestyle="--",
        linewidth=2,
        label=f"Lower Bound = {lower_bound:.2f}"
    )

    plt.axvline(
        upper_bound,
        linestyle="--",
        linewidth=2,
        label=f"Upper Bound = {upper_bound:.2f}"
    )

    plt.axvline(
        data.median(),
        linestyle="-",
        linewidth=2,
        label=f"Median = {data.median():.2f}"
    )

    plt.title(
        f"{col} Distribution with Outlier Limits",
        fontsize=16,
        fontweight="bold"
    )

    plt.xlabel(col)
    plt.ylabel("Frequency")

    plt.legend()
    plt.grid(axis="y", alpha=0.3)

    plt.text(
        0.98,
        0.90,
        f"Outliers: {len(outliers)}",
        transform=plt.gca().transAxes,
        ha="right",
        fontsize=11,
        bbox=dict(
            boxstyle="round",
            facecolor="white",
            edgecolor="black"
        )
    )

    plt.tight_layout()
    plt.show()
```

```python
for col in numeric_col:
    skewness = df[col].skew()
    
    print(f"{col}: Skewness = {skewness:.3f}")
```
